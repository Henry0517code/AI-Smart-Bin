This notebook covers how to run inference with a YOLO image classification model and how to read the results.

## 1. Load a Model

Ultralytics provides pretrained classification models trained on ImageNet-1k.
The model name ends in `-cls` to distinguish it from detection models.

| Model | Size | ImageNet Accuracy |
|-------|------|-------------------|
| `yolo11n-cls.pt` | Nano (smallest, fastest) | ~70% |
| `yolo11s-cls.pt` | Small | ~74% |
| `yolo11m-cls.pt` | Medium | ~77% |
| `yolo11l-cls.pt` | Large | ~78% |
| `yolo11x-cls.pt` | XLarge (most accurate) | ~79% |

````python
from ultralytics import YOLO

model = YOLO("yolo11n-cls.pt")  # Downloads automatically on first run
````

## 2. Run Inference

`model()` accepts:
- A file path: `"image.jpg"`
- A URL: `"https://..."`
- A numpy array or PIL Image
- A list of any of the above (batch inference)

It always returns a **list** of `Results` objects — one per input image.

````python
results = model("https://ultralytics.com/images/bus.jpg")

print(type(results))  # <class 'list'>
print(type(results[0]))  # <class 'ultralytics.engine.results.Results'>
````

## 3. The `Results` Object

Each `Results` object has a `.probs` attribute — a `Probs` object that holds the classification output.

```
Results
├── .orig_img        — original image as a numpy array
├── .names           — dict mapping class ID → class name
└── .probs           — Probs object
    ├── .data        — raw probability tensor for all classes  (shape: [num_classes])
    ├── .top1        — class ID of the top prediction          (int)
    ├── .top1conf    — confidence score of the top prediction  (float)
    └── .top5        — class IDs of the top-5 predictions      (list of ints)
```

````python
result = results[0]  # First (and only) image

probs = result.probs
print(probs)  # summary printout from ultralytics
````

## 4. Top-1 Prediction

The single most likely class.

````python
top1_id   = result.probs.top1  # e.g. 874
top1_conf = result.probs.top1conf  # e.g. tensor(0.87)
top1_name = result.names[top1_id]  # e.g. 'school bus'

print(f"Prediction : {top1_name}")
print(f"Confidence : {top1_conf:.2%}")
````

## 5. Top-5 Predictions

`.top5` gives the five highest-scoring class IDs. Pair them with `.probs.data` to get each confidence.

````python
for rank, class_id in enumerate(result.probs.top5, start=1):
    name = result.names[class_id]
    confidence = result.probs.data[class_id].item()
    print(f"#{rank}  {name:<25} {confidence:.2%}")
````

## 6. Live Webcam Capture

So far we downloaded an image from the internet. But you can also **take a photo live from your webcam** and send it straight to YOLO.

We use **OpenCV** (`cv2`) for this. OpenCV is a popular computer vision library that can, among many things, talk to your camera hardware.

Here is what the code does step by step:

1. `cv2.VideoCapture(0)` — opens the first camera it finds (index `0` = your built-in webcam)
2. `cap.read()` — grabs one frame; returns a flag (did it work?) and the image as a NumPy array
3. `cv2.cvtColor(...)` — OpenCV stores colours as **BGR** (Blue-Green-Red), but PIL expects **RGB**, so we swap the channels
4. `Image.fromarray(...)` — converts the NumPy array into a PIL Image, which YOLO also accepts
5. `cap.release()` — always close the camera when you're done so it's free for other apps

````python
%pip install opencv-python
````

````python
import cv2
from PIL import Image

cap = cv2.VideoCapture(0)  # 0 = default webcam
ok, frame = cap.read()     # Grab one frame
cap.release()              # Done with the camera

if not ok:
    raise RuntimeError("Could not read from webcam — is it connected?")

# OpenCV gives BGR, convert to RGB for PIL / YOLO
frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
webcam_image = Image.fromarray(frame_rgb)

webcam_image   # Preview in Jupyter
````

````python
# Run YOLO classification on the webcam frame
results = model(webcam_image)
result  = results[0]

print(f"Prediction : {result.names[result.probs.top1]}")
print(f"Confidence : {result.probs.top1conf:.2%}")
````
